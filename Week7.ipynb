**Average Filter**

```python
# Baca gambar dan ubah ke grayscale
img = cv.imread('/content/drive/MyDrive/images/female.jpeg')

# Check if the image was loaded successfully
if img is None:
    print("Error: Could not load image. Please check the file path.")
else:
    gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

    # Fungsi konvolusi manual
    def convolution(image, kernel):
        h, w = image.shape
        kh, kw = kernel.shape
        # Calculate padding size
        pad_h = kh // 2
        pad_w = kw // 2
        # Apply padding
        img_padded = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='constant')
        output = np.zeros((h, w))
        for y in range(h):
            for x in range(w):
                # Extract region from padded image
                region = img_padded[y:y+kh, x:x+kw]
                output[y, x] = np.sum(region * kernel)
        return np.clip(output, 0, 255).astype(np.uint8)


    # Kernel Average Filter
    kernel_avg = np.ones((3,3)) / 9

    # Terapkan konvolusi
    avg_result = convolution(gray, kernel_avg)

    # Tampilkan hasil
    plt.figure(figsize=(10,5))
    plt.subplot(1,2,1)
    plt.imshow(gray, cmap='gray')
    plt.title('Original')
    plt.axis('off')

    plt.subplot(1,2,2)
    plt.imshow(avg_result, cmap='gray')
    plt.title('Average Filter')
    plt.axis('off')

    plt.show()
```
